**Problem 1: Social Media – Post Engagement Analyzer**


```python
def analyze_engagement(total_likes):
    print(f"Total Likes: {total_likes}")
    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")
analyze_engagement(1050)
```

**Problem 3: Agriculture – Rainfall Adequacy Checker**


```python
def analyze_rainfall(rainfall_data, required_level):
    total_rainfall = 0
    for rain in rainfall_data:
        total_rainfall += rain

    # Calculate average
    average_rainfall = total_rainfall / len(rainfall_data)

    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print(f"Average Rainfall: {int(average_rainfall)}")
    print(f"Rainfall Status: {status}")

#Example usage
seasonal_rainfall = [70, 75, 72, 71, 73]
required_rainfall_level = 70
analyze_rainfall(seasonal_rainfall, required_rainfall_level)
```

**Problem 4: Social Media – Duplicate Account Detection**

```python
def check_duplicates(usernames):
    if len(usernames) != len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

# Example usage
usernames_list = ["alice", "bob", "charlie", "alice"]
check_duplicates(usernames_list)
```

**Problem 5: Healthcare – Appointment Eligibility Checker**

```python
def check_eligibility(age):
    print(f"Patient Age: {age}")
    if age >= 18:
        print("Eligibility Status: Eligible")
    else:
        print("Eligibility Status: Not Eligible")

# Example usage
check_eligibility(21)
```

**Problem 6: Agriculture – Premium Crop Price Filter**


```python
def premium_crops(prices):
    # Filter prices above 2000
    premium = [price for price in prices if price > 2000]
    print(f"Premium Crops: {premium}")

# Example usage
crop_prices = [1500, 2500, 1800, 3200, 2000]
premium_crops(crop_prices)
```
