Problem 1: Unique Words in a Sentence

```python
sentence = input("Enter thr sentence: ")
words = sentence.split()
unique_words = set(words)
print("Unique words:", unique_words)
print("Count of unique words:", len(unique_words))
```

Problem 2: Highest Salary from Employee Data


```python
employees = {"Ravi": 75000,"Anita": 68000,"Kiran": 72000}
highest_employee = max(employees,key=employees.get)
highest_salary = employees[highest_employee]
print(f"Highest salary: { highest_salary} - {highest_employee}")
```

Problem 4: Count Products Above a Price Threshold


```python
prices = [450, 1200, 899, 1500, 300]
count = 0
for prc in prices:
  if prc > 1000:
    count = count + 1
print("Number of products above 1000:", count)
```

Problem 5: Calculate Attendance Percentage

```python
attendance = ["P", "P", "A", "P", "P"]
present_days= attendance.count("P")
total_days = len(attendance)
attendance_percentage = (present_days/total_days)*100
print('Attendence Percentage: ',attendance_percentage)
```

Problem 7: Count Character Frequency


```python
text = "pythonp"
char_count = {}
for ch in text:
    if ch in char_count:
        char_count[ch] += 1
    else:
        char_count[ch] = 1
print(char_count)
```
