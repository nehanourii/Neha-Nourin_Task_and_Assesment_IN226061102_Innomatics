Problem Statement 1: Employee Performance Bonus Eligibility


```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
    }
# Find highest score
highest_score = max(employees.values())

# Get employees with highest score
top_performers = [name for name, score in employees.items() if score == highest_score]
print(f"Top performers Eligible for Bonus: {', '.join(top_performers)} (Score: {highest_score})")
```

Problem Statement 2: Search Query Keyword Analysis


```python
# Input search query
query = "Buy mobile phone buy phone online"
query = query.lower()
words = query.split()

# Count frequency of each word
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Store only repeated words
result = {}
for word in frequency:
    if frequency[word] > 1:
        result[word] = frequency[word]
print(result)
```

Problem Statement 3: Sensor Data Validation

```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
# Store valid readings
valid_readings = []
for index in range(len(sensor_readings)):
    if sensor_readings[index] % 2 == 0:

 # Store (hour, value) as tuple
        valid_readings.append((index, sensor_readings[index]))
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

Problem Statement 4: Email Domain Usage Analysis


```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]
total = len(emails)
# Extract domain names from emails
domains = [email.split("@")[1] for email in emails]
# Calculate and print percentage for each unique domain
for domain in set(domains):
    percent = (domains.count(domain) / total) * 100
    print(f"{domain}: {int(percent)}%")
```

Problem Statement 5: Sales Spike Detection

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
# Calculate average sales
avg = sum(sales) / len(sales)
# Check for sales more than 30% above average
for i in range(len(sales)):
    if sales[i] > avg * 1.30:
        print(f"Day {i+1}: {sales[i]}")
```

Problem Statement 6: Duplicate User ID Detection

```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
count_dict = {}
# Count occurrences
for user in user_ids:
    if user in count_dict:
        count_dict[user] += 1
    else:
        count_dict[user] = 1
#Print only duplicates
for user, count in count_dict.items():
    if count > 1:
        print(f"{user} → {count} times")
```
