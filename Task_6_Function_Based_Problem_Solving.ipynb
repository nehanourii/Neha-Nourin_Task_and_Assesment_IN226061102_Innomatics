**Problem Statement 1:Smart Parking Lot Management System**

```python
def smart_parking_system(capacity, logs):
    parked = 0
    peak_usage = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1

        # Track peak usage
        if parked > peak_usage:
            peak_usage = parked

        # Alert if capacity exceeded
        if parked > capacity:
            print("⚠ Alert: Parking capacity exceeded!")

    print("Currently Parked Vehicles:", parked)

    if parked < capacity:
        print("Parking Status: Available")
    else:
        print("Parking Status: Full")

    print("Peak Parking Usage:", peak_usage)

capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking_system(capacity, logs)
```


**Problem Statement 2:Online Food Delivery Time Estimator**

```python
def estimate_delivery_time(distance, traffic, weather):
    # Base speed assumption (km per hour)
    base_speed = 30

    base_time = (distance / base_speed) * 60

    if traffic == "High":
        base_time += 15
    elif traffic == "Medium":
        base_time += 8
    elif traffic == "Low":
        base_time += 3

    if weather == "Rainy":
        base_time += 10
    elif weather == "Storm":
        base_time += 20
    elif weather == "Clear":
        base_time += 0

    print("Estimated Delivery Time:", round(base_time), "minutes")

estimate_delivery_time(8, "High", "Rainy")
```


**Problem Statement 3:Movie Theatre Seat Occupancy Analyzer**


```python
def analyze_seat_booking(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100

    print("Occupancy:", round(occupancy), "%")
    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")

    # Suggest additional shows
    if occupancy >= 90:
        print("Suggestion: Consider opening additional shows.")

total_seats = 200
booked_seats = [1] * 150

analyze_seat_booking(total_seats, booked_seats)
```


 **Problem Statement 4:Cloud Server Load Classification System**

```python
def classify_server_load(cpu_readings):
    total = 0
    for reading in cpu_readings:
        total += reading
    average = total / len(cpu_readings)

    print("Average CPU Load:", round(average), "%")
    if average < 50:
        print("Server Status: Normal")
    elif 50 <= average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

cpu_data = [45, 60, 70, 85, 90]
classify_server_load(cpu_data)
```

**Problem Statement 5:Smart Classroom Resource Usage Monitor**

```python
def track_resource_usage(resource_usage):
    overused = []
    # Define overuse limit (hours per day)
    limit = 8

    for resource, hours in resource_usage.items():
        if hours > limit:
            overused.append(resource)
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

usage_data = {"Projector": 6,
    "AC": 9, "Lights": 4}

track_resource_usage(usage_data)
```


 **Problem Statement 6: Online Event Registration Capacity Controller**


```python
def manage_event_registration(capacity, total_registrations):
    confirmed = 0
    waitlisted = 0

    for i in range(total_registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    if confirmed >= capacity:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")

manage_event_registration(100, 105)
```
