**Problem 1: Mobile Recharge Validation System**

```python
def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]
    if amount < 50:
        print("Recharge amount must be at least ₹50.")
        return False
    elif amount not in valid_plans:
        print("Invalid plan selected.")
        print("Available plans:", valid_plans)
        return False
    else:
        print("Recharge Successful!")
        return True

while True:
    try:
        amount = int(input("Enter recharge amount: ₹"))
        if validate_recharge(amount):
            break   # Exit loop if valid
        print("Please try again")
    except ValueError:
        print("Please enter a valid numeric amount")
```

```python
def check_stock(products):
    for product, quantity in products.items():
        if quantity < 15:
            print(f"{product} : {quantity}  - Reorder Alert")
        else:
            print(f"{product} : {quantity} -  Stock OK")
product_stock = {"Apple": 10,"Banana": 30,"Orange": 5,"Tomato": 20}
check_stock(product_stock)
```

**Problem 3: Student Result Processing System**


```python
def check_result(marks_list):
    total = 0
    for mark in marks_list:
        total += mark
    average = total / len(marks_list)

    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"
    return average, result
marks = [81, 77, 91, 64, 75]
average, result = check_result(marks)

print("Average Marks:", average)
print("Result:", result)
```

**Problem 4: Cab Fare Estimator with Retry Option**


```python
def calculate_fare(distance, peak):
    base_fare = 50
    per_km_rate = 12

    fare = base_fare + (distance * per_km_rate)

    if peak.lower() == "yes":
        fare = fare + (fare * 0.25)   # 25% extra

    return fare

while True:
    try:
        distance = float(input("Enter distance in km: "))
        peak = input("Is it peak hour? (yes/no): ")

        total_fare = calculate_fare(distance, peak)
        print("Total Fare: ₹", total_fare)

        retry = input("Do you want to calculate again? (yes/no): ")
        if retry.lower() != "yes":
            print("Thank you for using Cab Fare Calculator")
            break

    except ValueError:
        print("Please enter valid numeric distance")
```

**Problem 5: Employee Attendance Eligibility Checker**

```python
def check_attendance(attendance_list):
    present_count = 0

    for status in attendance_list:
        if status == "P":
            present_count += 1

    total_days = len(attendance_list)
    percentage = (present_count / total_days) * 100

    if percentage >= 75:
        return percentage, "Eligible"
    else:
        return percentage, "Not Eligible"

attendance = input("Enter attendance separated by space (P/A): ").split()

percent, result = check_attendance(attendance)

print("Attendance Percentage:", percent)
print("Eligibility Status:", result)
```

**Problem 6: Password Strength Checker**


```python
def check_password(password):
    has_digit = False
    has_special = False
    if len(password) < 8:
        return "Weak Password(Minimum 8 characters required)"
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in "@#$":
            has_special = True

    if has_digit and has_special:
        return "Strong Password "
    else:
        return "Weak Password(Must include digit and @#$)"
pwd = input("Enter your password: ")
result = check_password(pwd)
print(result)
```
